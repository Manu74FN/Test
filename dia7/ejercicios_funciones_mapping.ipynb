##### 1. Calcula la diferencia entre el máximo y el mínimo por fila y por columna aplicando una función al dataframe df

```python
df = pd.DataFrame({'A': [1,2,3,4]
                   , 'B': [10,20,30,40]
                   , 'C': [20,40,60,80]
                   },
                   index=['Row 1', 'Row 2','Row 3', 'Row 4'])
df
```

```python
def diferencia (x):
    return x.max() - x.min()


df['Dif']= df.apply(diferencia, axis=1)
df
```

```python
df.loc['Dife'] = df.apply(lambda x:x.max()-x.min(), axis=0)
df
```

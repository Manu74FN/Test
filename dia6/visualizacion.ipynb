### Visualización básica con pandas, basado en su propia documentación

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
air_quality = pd.read_csv("/home/manu/Desktop/Test/data/air_quality_no2.csv", index_col=0, parse_dates= True)
air_quality.head()
```

##### El método plot de pandas, nos permite hacer gráficas de Series y Dataframes

```python
air_quality.plot()
```

##### Podemos graficar columnas individualmente

```python
air_quality["station_paris"].plot()
```

##### Podemos crear distintos tipos de gráficas con los distintos tipos que tiene plot.
##### Aquí vemos un listado con todos los tipos de gráficos soportados.

```python
[
    method_name
    for method_name in dir(air_quality.plot)
    if not method_name.startswith("_")
]
```

##### Para ver la dispersión de una variable frente a otra, usamos el gráfico scatter

```python
air_quality.plot.scatter(x= "station_london", y= "station_paris", alpha= 0.5)
```

```python
air_quality.plot.box()
```

##### Podemos hacer una gráfica separada para cada una de las columnas

```python
axs = air_quality.plot.area(figsize=(12, 4), subplots=True)
plt.show
```

##### Los gráficos se pueden personalizar y salvar a un fichero.
##### Cada uno de los gráficos es un objeto de matplotlib.
##### Matplotlib en la librería más usada para hacer gráficos en python, aunque no la única

```python
fig, axs = plt.subplots(figsize=(12, 4))
air_quality.plot.area(ax = axs)
axs.set_ylabel("NO$_2$ concentracion")
fig.savefig("fig7no2_concentracion.png")
```

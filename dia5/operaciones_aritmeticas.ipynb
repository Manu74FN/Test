### Operaciones aritméticas

##### Las operaciones se aplican cuando coinciden los índices en ambas estructuras(series , dataframe).
##### Si coinciden los índices rellenacon NaN.
##### Pandas tiene varias funciones básicas que permiten establecer un valor de relleno
##### En un numpy, los arrays deben tener las mismas dimensiones para realizar las operaciones

```python
import numpy as np
import pandas as pd
```

##### Sumamos dos series donde coinciden los índices

```python
serie1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
serie2 = pd.Series([10, 11, 12], index=['a', 'b', 'c'])
serie1+serie2
```

##### Ahora sumamos dos series donde no coinciden los índices

```python
serie1 = pd.Series([0, 1, 2, 3,], index=['a', 'b', 'k', 'l'])
serie2 = pd.Series([10, 11, 12, 13, 14, 15, 16], index=['a', 'b', 'c', 'd', 'f', 'g', 'h'])
serie1+serie2
```

##### Ejemplos con Dataframe

```python
lista_pares = [x for x in range(10) if x % 2 == 0]
lista_impares = [x for x in range(10) if x % 2 != 0]
print(lista_pares)
print(lista_impares)
```

```python
pares_df = pd.DataFrame(index=['a', 'b', 'c', 'd', 'e'])
impares_df = pd.DataFrame(index=['a', 'b', 'c', 'd', 'f'])
pares_df["val"] = lista_pares
impares_df["val"] = lista_impares
print("Pares:\n", pares_df)
print("Impares:\n", impares_df)
```

##### Sumamos los dos dataframe

```python
pares_df.add(impares_df)
```

##### Ahora los sumamos, pero dando un valor a los nulos

```python
pares_df.add(impares_df, fill_value=0)
```

##### Ahora los multiplicamos

```python
pares_df.mul(impares_df, fill_value=0)
```

```python
data1 = pd.DataFrame([10, 20, 30], index=['a', 'b', 'c'])
data2 = pd.DataFrame([5, 10, 15], index=['a', 'b', 'c'])
print("data1:\n", data1)
print("data2:\n", data2)
```

##### Aquí la diferencia entre div y rdiv

```python
data1.div(data2)
```

```python
data1.rdiv(data2)
```

##### Diferencia entre sub y rsub

```python
data1.sub(data2)
```

```python
data1.rsub(data2)
```
